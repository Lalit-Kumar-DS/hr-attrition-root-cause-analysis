```python
#handeling null values
```

```python
employees.isnull().sum()
```

```python
employees['environment_satisfaction'] = employees['environment_satisfaction'].fillna(employees['environment_satisfaction'].median())
employees['job_satisfaction'] = employees['job_satisfaction'].fillna(employees['job_satisfaction'].median())
employees['work_life_balance'] = employees['work_life_balance'].fillna(employees['work_life_balance'].median())
```

```python
employees = employees.astype({'environment_satisfaction': 'int',	'job_satisfaction': 'int',	'work_life_balance': 'int'})
```

```python
import psycopg2
from sqlalchemy import create_engine
import pandas as pd

# Connect to PostgreSQL
engine = create_engine("postgresql+psycopg2://postgres:1234@localhost:5432/hr_analytics")

#load to postgres

general_data.to_sql('general_data', engine, if_exists='replace', index=False)
employees.to_sql('employees', engine, if_exists='replace', index=False)
manager_sur.to_sql('manager_sur', engine, if_exists='replace', index=False)


# Test query
df = pd.read_sql("SELECT * FROM general_data LIMIT 5;", engine)
print(df)
```
